```python
from kafka import KafkaProducer
import pandas as pd
import json
import random
import time
from datetime import datetime
# ----------------------------------
# Kafka Configuration
# ----------------------------------
producer = KafkaProducer(bootstrap_servers="ed-kafka:29092",value_serializer=lambda x:json.dumps(x).encode("utf-8"))
print("Kafka Producer Connected Successfully")
# ----------------------------------
# Load Product Dataset
# ----------------------------------
products = pd.read_csv("../dataset/products.csv")
print("Products Loaded:",len(products))
# ----------------------------------
# Master Data
# ----------------------------------
cities = ["Pune","Mumbai","Delhi","Bangalore","Hyderabad","Chennai"]
payment_methods = ["UPI","Credit Card","Debit Card","Cash"]
order_status = ["Placed","Processing","Delivered","Cancelled"]
# ----------------------------------
# Generate Order Function
# ----------------------------------
def generate_order():
    product = products.sample(1).iloc[0]
    quantity = random.randint(1,5)
    order = {
        "order_id":f"ORD{random.randint(100000,999999)}",
        "customer_id":f"CUST{random.randint(1000,9999)}",
        "product_id":product["product_id"],
        "product_name":product["product_name"],
        "category":product["category"],
        "sub_category":product["sub_category"],
        "brand":product["brand"],
        "quantity":quantity,
        "price":float(product["price"]),
        "total_amount":float(quantity *product["price"]),
        "city":random.choice(cities),
        "payment_method":random.choice(payment_methods),
        "order_status":random.choice(order_status),
        "order_time":datetime.now().strftime("%Y-%m-%d %H:%M:%S")}
    return order
# ----------------------------------
# Send Messages To Kafka
# ----------------------------------
while True:
    order = generate_order()
    producer.send("ecommerceOrders",value=order)
    print("Order Sent:",order)
    time.sleep(0)
```
